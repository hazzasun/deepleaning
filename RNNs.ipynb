```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, GRU, SimpleRNN, Dense, Dropout
from tensorflow.keras.optimizers import Adam

# ----------------------------------------------------------------
# Parameters
# ----------------------------------------------------------------
window_size = 30
epochs = 50
batch_size = 32
learning_rate = 0.001
train_ratio = 0.7
val_ratio = 0.2

# ----------------------------------------------------------------
# 1. Data Loading and Preprocessing
# ----------------------------------------------------------------
# Load datasets
train_df = pd.read_csv('Google_Stock_Price_Train.csv')
test_df = pd.read_csv('Google_Stock_Price_Test.csv')

# Fix data by ensuring columns are strings and then removing commas
for col in ['Open', 'High', 'Low', 'Close', 'Volume']:
    train_df[col] = train_df[col].astype(str).str.replace(',', '').astype(float)
    test_df[col] = test_df[col].astype(str).str.replace(',', '').astype(float)

# Extract features
feature_columns = ['Open', 'High', 'Low', 'Close', 'Volume']
train_data = train_df[feature_columns].values
test_data = test_df[feature_columns].values

# Scale data to range [0, 1]
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_train_data = scaler.fit_transform(train_data)

# Train/Val Split
train_size = int(len(scaled_train_data) * train_ratio)
val_size = int(len(scaled_train_data) * val_ratio)

train_subset = scaled_train_data[:train_size]
val_subset = scaled_train_data[train_size:train_size + val_size]
remaining_train_subset = scaled_train_data[:train_size + val_size]

# Create time-series datasets
def create_dataset(dataset, window_size=30):
    X, y = [], []
    for i in range(window_size, len(dataset)):
        X.append(dataset[i - window_size:i, :])
        y.append(dataset[i, :])
    return np.array(X), np.array(y)

X_train, y_train = create_dataset(train_subset, window_size)
X_val, y_val = create_dataset(val_subset, window_size)

# Prepare Test Set
full_data_for_test = np.concatenate((remaining_train_subset, scaler.transform(test_data)), axis=0)
X_test, y_test_scaled = create_dataset(full_data_for_test, window_size)

# Ensure that y_test_scaled corresponds to the predictions
y_true = scaler.inverse_transform(y_test_scaled)

# ----------------------------------------------------------------
# 2. Model Function
# ----------------------------------------------------------------
def build_model(model_type, input_shape):
    model = Sequential()
    if model_type == 'VanillaRNN':
        model.add(SimpleRNN(units=100, return_sequences=True, input_shape=input_shape))
        model.add(Dropout(0.2))
        model.add(SimpleRNN(units=100))
        model.add(Dropout(0.2))
    elif model_type == 'GRU':
        model.add(GRU(units=100, return_sequences=True, input_shape=input_shape))
        model.add(Dropout(0.2))
        model.add(GRU(units=100))
        model.add(Dropout(0.2))
    elif model_type == 'LSTM':
        model.add(LSTM(units=100, return_sequences=True, input_shape=input_shape))
        model.add(Dropout(0.2))
        model.add(LSTM(units=100))
        model.add(Dropout(0.2))
    else:
        raise ValueError("Invalid model type. Choose from 'VanillaRNN', 'GRU', 'LSTM'.")
    model.add(Dense(units=len(feature_columns)))
    return model

# ----------------------------------------------------------------
# 3. Training and Evaluation
# ----------------------------------------------------------------
results = []

for model_type in ['VanillaRNN', 'GRU', 'LSTM']:
    print(f"Training {model_type} model...")
    model = build_model(model_type, (X_train.shape[1], X_train.shape[2]))

    optimizer = Adam(learning_rate=learning_rate)
    model.compile(loss='mean_squared_error', optimizer=optimizer)

    history = model.fit(X_train, y_train,
                        epochs=epochs,
                        batch_size=batch_size,
                        validation_data=(X_val, y_val),
                        shuffle=False,
                        verbose=1)

    # Predict on test set
    y_pred_scaled = model.predict(X_test)
    y_pred = scaler.inverse_transform(y_pred_scaled)

    # Compute metrics
    mse = mean_squared_error(y_true, y_pred)
    mape = mean_absolute_percentage_error(y_true, y_pred)

    print(f"{model_type} - Test MSE: {mse}, Test MAPE: {mape}")
    results.append({'Model': model_type, 'MSE': mse, 'MAPE': mape})

    # Plot training history
    plt.figure(figsize=(10, 5))
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title(f'{model_type} Training & Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('MSE Loss')
    plt.legend()
    plt.show()

# ----------------------------------------------------------------
# 4. Comparison Table
# ----------------------------------------------------------------
results_df = pd.DataFrame(results)
print("\nComparison of Models:")
print(results_df)

# Plot MSE Comparison
plt.figure(figsize=(8, 5))
plt.bar(results_df['Model'], results_df['MSE'], color='skyblue')
plt.title('MSE Comparison Between Models')
plt.ylabel('Mean Squared Error')
plt.show()

# Plot MAPE Comparison
plt.figure(figsize=(8, 5))
plt.bar(results_df['Model'], results_df['MAPE'], color='lightgreen')
plt.title('MAPE Comparison Between Models')
plt.ylabel('Mean Absolute Percentage Error')
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error

# ------------------------------------------------------------
# 1. Data Loading and Preprocessing
# ------------------------------------------------------------
# Load the dataset
train_df = pd.read_csv('Google_Stock_Price_Train.csv')
test_df = pd.read_csv('Google_Stock_Price_Test.csv')

# Define the columns we are interested in
feature_columns = ['Open', 'High', 'Low', 'Close', 'Volume']

# Fixing potential string-to-float issues by removing commas where necessary
for col in feature_columns:
    if train_df[col].dtype == 'object':
        train_df[col] = train_df[col].str.replace(',', '').astype(float)
    if test_df[col].dtype == 'object':
        test_df[col] = test_df[col].str.replace(',', '').astype(float)

# Extract and scale the features from the training set
train_data = train_df[feature_columns].values
test_data = test_df[feature_columns].values

scaler = MinMaxScaler(feature_range=(0, 1))
scaled_train_data = scaler.fit_transform(train_data)

# ------------------------------------------------------------
# 2. Train/Validation Split
# ------------------------------------------------------------
# Let’s say we use 70% of the training data for actual training, and 20% for validation.
train_ratio = 0.7
val_ratio = 0.2

train_size = int(len(scaled_train_data) * train_ratio)
val_size = int(len(scaled_train_data) * val_ratio)

train_subset = scaled_train_data[:train_size]
val_subset = scaled_train_data[train_size:train_size + val_size]
remaining_train_subset = scaled_train_data[:train_size + val_size]

# Create input-output pairs for train, val, and test datasets
def create_dataset(dataset, window_size=30):
    X, y = [], []
    for i in range(window_size, len(dataset)):
        X.append(dataset[i-window_size:i, :])  # past 30 days
        y.append(dataset[i, :])  # predict the next day (all 5 features)
    return np.array(X), np.array(y)

X_train, y_train = create_dataset(train_subset)
X_val, y_val = create_dataset(val_subset)

# ------------------------------------------------------------
# 3. Prepare Test Set
# ------------------------------------------------------------
# For the test set, combine the last 30 days of training data with test data for context.
full_data_for_test = np.concatenate((remaining_train_subset, scaler.transform(test_data)), axis=0)
X_test, y_test = create_dataset(full_data_for_test)

# ------------------------------------------------------------
# 4. Build the LSTM Model
# ------------------------------------------------------------
model = Sequential()
model.add(LSTM(units=100, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dropout(0.2))
model.add(LSTM(units=100))
model.add(Dropout(0.2))
model.add(Dense(units=len(feature_columns)))  # predict all 5 features

optimizer = Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error', optimizer=optimizer)

model.summary()

# ------------------------------------------------------------
# 5. Train the Model
# ------------------------------------------------------------
history = model.fit(X_train, y_train,
                    epochs=50,
                    batch_size=32,
                    validation_data=(X_val, y_val),
                    shuffle=False)

# ------------------------------------------------------------
# 6. Evaluation on Test Set
# ------------------------------------------------------------
# Predictions on test set
y_pred_scaled = model.predict(X_test)

# Invert scaling to get predictions and ground truth in original scale
y_pred = scaler.inverse_transform(y_pred_scaled)
y_true = scaler.inverse_transform(y_test)

# ------------------------------------------------------------
# 7. Compute Evaluation Metrics
# ------------------------------------------------------------
mse = mean_squared_error(y_true, y_pred)
mape = mean_absolute_percentage_error(y_true, y_pred)

print("Test MSE:", mse)
print("Test MAPE:", mape)

# ------------------------------------------------------------
# 8. Visualize Predictions vs True Values
# ------------------------------------------------------------
plt.figure(figsize=(12, 6))
plt.plot(y_true[:, 3], label='True Close Price', color='blue')  # Index 3 corresponds to 'Close'
plt.plot(y_pred[:, 3], label='Predicted Close Price', color='orange')
plt.title('Google Stock Price Prediction (Close)')
plt.xlabel('Time Step')
plt.ylabel('Price')
plt.legend()
plt.show()

# ------------------------------------------------------------
# 9. Save the Model
# ------------------------------------------------------------
model.save("lstm_stock_model.h5")

# ------------------------------------------------------------
# 10. Plot Training and Validation Loss
# ------------------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training & Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.legend()
plt.show()
```
